## Relationship Between `X`, `Y`, and `Z`

The RadioML 2018 dataset stores the IQ waveform, modulation label, and Signal-to-Noise Ratio (SNR) in three separate datasets. These datasets are synchronized using a common sample index. Therefore, the *i-th* entry of `X`, `Y`, and `Z` together represent one complete radio signal instance.

For example:

| Sample Index | IQ Signal (`X`) | Modulation Label (`Y`) | SNR (`Z`) |
|:------------:|:----------------|:-----------------------|:---------:|
| **0** | `X[0]` → IQ waveform | `Y[0]` → Modulation type | `Z[0]` → SNR |
| **1** | `X[1]` → IQ waveform | `Y[1]` → Modulation type | `Z[1]` → SNR |
| **2** | `X[2]` → IQ waveform | `Y[2]` → Modulation type | `Z[2]` → SNR |
| ⋮ | ⋮ | ⋮ | ⋮ |
| **N** | `X[N]` → IQ waveform | `Y[N]` → Modulation type | `Z[N]` → SNR |

### Key Concept

For every sample index **i**:

- `X[i]` contains the IQ signal samples.
- `Y[i]` contains the corresponding modulation class.
- `Z[i]` contains the corresponding Signal-to-Noise Ratio (SNR).

These three entries always refer to the **same transmitted signal**.

For example, if `i = 150`:

```text
X[150]  → IQ waveform
Y[150]  → QPSK
Z[150]  → 10 dB
```

This means that the IQ waveform stored in `X[150]` was generated using **QPSK modulation** and corrupted by noise corresponding to an **SNR of 10 dB**.

```text
                    Sample Index = i
                           │
        ┌──────────────────┼──────────────────┐
        │                  │                  │
        ▼                  ▼                  ▼
      X[i]              Y[i]               Z[i]
 IQ Waveform      Modulation Label      SNR Value
        │                  │                  │
        └──────────────────┴──────────────────┘
                           │
                           ▼
                One Radio Signal Instance
```

## 6. Dataset Index Explorer

Each RadioML sample is identified by a single dataset index `i`.

For a given index:

- `X[i]` contains the 1024 × 2 I/Q samples.
- `Y[i]` contains the one-hot encoded modulation label.
- `Z[i]` contains the SNR value.

Therefore, each index can be represented as:

    i → (IQ signal, modulation, SNR)

The purpose of this section is to explore the relationship between dataset indices, modulation classes, and SNR values.

This will allow us to select specific samples such as:

- OOK at 10 dB
- BPSK at 0 dB
- QPSK at -10 dB
- AM-DSB-WC at 20 dB

````python
import pandas as pd
'''this code snippet is used to create a distribution table of modulation types across different SNR values. It uses the pandas library to create a crosstab,
 which shows the count of samples for each modulation type at each SNR level. The resulting distribution table can be useful for analyzing the dataset and 
 understanding how different modulation types are represented across various SNR conditions.'''

modulation_names = np.array(classes)[labels]

distribution = pd.crosstab(
    modulation_names,
    snrs
)

distribution
````

````python
'''this function returns the indices of samples that match a specific modulation type and SNR value. 
It takes two parameters: modulation (the desired modulation type) and snr (the desired SNR value). 
The function first finds the index of the specified modulation type in the classes list. 
Then, it uses NumPy's where function to find the indices of samples that have both the specified modulation type and SNR value. 
Finally, it returns the array of matching indices.
'''
def get_indices(modulation, snr):
    
    class_index = classes.index(modulation)

    indices = np.where(
        (labels == class_index) &
        (snrs == snr)
    )[0]

    return indices
````

````python
'''this code snippet demonstrates how to use the get_indices function to retrieve the indices of 
samples that match a specific modulation type and SNR value.'''

'''for example, it retrieves the indices of samples with modulation type "OOK" and SNR value 10.
The resulting indices are then printed, along with the total number of samples found, the first 10 indices, 
and the last 10 indices. This allows for a quick inspection of the samples that meet the specified criteria.'''

indices_ook = get_indices("OOK", 10)    

print("Number of samples:", len(indices_ook))
print("First 10 indices:", indices_ook[:10])
print("Last 10 indices:",indices_ook[-10:])
````

````python
'''this function selects a sample index based on the specified modulation type and SNR value. 
It takes three parameters: modulation (the desired modulation type), snr (the desired SNR value ),  
and random (a boolean indicating whether to select a random sample or the first one).'''

def select_sample(modulation, snr, random=False, seed=42):

    indices = get_indices(modulation, snr)

    if len(indices) == 0:
        raise ValueError(
            f"No samples found for {modulation} at {snr} dB."
        )

    if random:
        rng = np.random.default_rng(seed)
        index = rng.choice(indices)
    else:
        index = indices[0]

    return index
````

````python
'''this function inspects a sample from the dataset based on the specified modulation type and SNR value.
It takes four parameters: modulation (the desired modulation type), snr (the desired SNR value),
random (a boolean indicating whether to select a random sample or the first one), and seed (the random 
seed for reproducibility).'''

def inspect_sample(modulation, snr, random=False, seed=42):

    index = select_sample(
        modulation,
        snr,
        random=True,
        seed=seed
    )

    iq = X[index]

    actual_label = classes[np.argmax(Y[index])]
    actual_snr = Z[index].item()

    print("Sample Information")
    print("-" * 35)
    print(f"Index      : {index}")
    print(f"Modulation : {actual_label}")
    print(f"SNR        : {actual_snr} dB")
    print(f"IQ Shape   : {iq.shape}")

    return index, iq
````

````python
def get_first_last_indices(labels, snrs, classes):
    """
    Return the first and last dataset index for every
    modulation type at every SNR level.
    here, labels-> array of label indices
    snrs-> array of SNR values
    classes-> list of modulation types
    """

    result = {}

    for class_index, modulation in enumerate(classes):

        result[modulation] = {}

        for snr in np.unique(snrs):

            indices = np.where(
                (labels == class_index) &
                (snrs == snr)
            )[0]

            if len(indices) > 0:
                result[modulation][snr] = {
                    "first": indices[0],
                    "last": indices[-1],
                    "count": len(indices)
                }

    return result
````

````python
index_ranges = get_first_last_indices(
    labels,
    snrs,
    classes
)
````

````python
for modulation in classes:

    print(f"\n{modulation}")
    print("-" * 35)

    for snr in sorted(index_ranges[modulation]):

        data = index_ranges[modulation][snr]

        print(
            f"{snr:>4} dB : "
            f"{data['first']} → {data['last']} "
            f"({data['count']} samples)"
        )
````

````python
def get_index_range(modulation, snr):
    """
    Return the first and last dataset indices for
    a given modulation and SNR.
    """

    if modulation not in index_ranges:
        raise ValueError(f"Unknown modulation: {modulation}")

    if snr not in index_ranges[modulation]:
        raise ValueError(
            f"No data found for {modulation} at {snr} dB"
        )

    first = index_ranges[modulation][snr]["first"]
    last = index_ranges[modulation][snr]["last"]

    return first, last
````
