## Relationship Between `X`, `Y`, and `Z`

The RadioML 2018 dataset stores the IQ waveform, modulation label, and Signal-to-Noise Ratio (SNR) in three separate datasets. These datasets are synchronized using a common sample index. Therefore, the *i-th* entry of `X`, `Y`, and `Z` together represent one complete radio signal instance.

For example:

| Sample Index | IQ Signal (`X`) | Modulation Label (`Y`) | SNR (`Z`) |
|:------------:|:----------------|:-----------------------|:---------:|
| **0** | `X[0]` → IQ waveform | `Y[0]` → Modulation type | `Z[0]` → SNR |
| **1** | `X[1]` → IQ waveform | `Y[1]` → Modulation type | `Z[1]` → SNR |
| **2** | `X[2]` → IQ waveform | `Y[2]` → Modulation type | `Z[2]` → SNR |
| ⋮ | ⋮ | ⋮ | ⋮ |
| **N** | `X[N]` → IQ waveform | `Y[N]` → Modulation type | `Z[N]` → SNR |

### Key Concept

For every sample index **i**:

- `X[i]` contains the IQ signal samples.
- `Y[i]` contains the corresponding modulation class.
- `Z[i]` contains the corresponding Signal-to-Noise Ratio (SNR).

These three entries always refer to the **same transmitted signal**.

For example, if `i = 150`:

```text
X[150]  → IQ waveform
Y[150]  → QPSK
Z[150]  → 10 dB
```

This means that the IQ waveform stored in `X[150]` was generated using **QPSK modulation** and corrupted by noise corresponding to an **SNR of 10 dB**.

```text
                    Sample Index = i
                           │
        ┌──────────────────┼──────────────────┐
        │                  │                  │
        ▼                  ▼                  ▼
      X[i]              Y[i]               Z[i]
 IQ Waveform      Modulation Label      SNR Value
        │                  │                  │
        └──────────────────┴──────────────────┘
                           │
                           ▼
                One Radio Signal Instance
```
