# Dataset Visualization

## Objective

The purpose of this notebook is to understand the RadioML 2018 dataset before developing the Automatic Modulation Classification (AMC) model.

This notebook focuses on:

- Understanding the dataset structure
- Inspecting individual samples
- Visualizing I/Q signals
- Understanding modulation labels
- Understanding Signal-to-Noise Ratio (SNR)

These visualizations will provide the intuition required before designing a deep learning model.

# Importing Libraries


```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
import json
```

# Loading Datasets

```python
DATASET_PATH = "E:\\AMC\\datasets\\RadioMl2018\\GOLD_XYZ_OSC.0001_1024.hdf5"

file = h5py.File(DATASET_PATH, "r")

X = file["X"]
Y = file["Y"]
Z = file["Z"]
```

# Imspecting one samples

```python
sample_idx=0

iq=X[sample_idx]
label=Y[sample_idx]
snr=Z[sample_idx]

print("IQ Shape :", iq.shape)
print()

print("First 5 IQ Samples")

print(iq[:5])

print()

print("Label")

print(label)

print()

print("SNR")

print(snr)
```

# Plotting I Component

```python
plt.figure(figsize=(12,4))

plt.plot(iq[:,0])  #iq[:,0] is the in-phase component meaning the real part of the complex signal.; iq[rows, columns]

plt.title("In-phase (I) Component")

plt.xlabel("Sample")

plt.ylabel("Amplitude")

plt.grid(True)

plt.show()
```

# Plotting the Q components

```python
plt.figure(figsize=(12,4))

plt.plot(iq[:,1])

plt.title("Quadrature (Q) Component")

plt.xlabel("Sample")

plt.ylabel("Amplitude")

plt.grid(True)

plt.show()
```

# IQ Constellation

```python
plt.figure(figsize=(6,6))

plt.scatter(
    iq[:,0],
    iq[:,1],
    s=5
)

plt.title("IQ Constellation")

plt.xlabel("I")

plt.ylabel("Q")

plt.grid(True)

plt.axis("equal")

plt.show()
```

# Signal Magnitude

```python
magnitude = np.sqrt(iq[:,0]**2 + iq[:,1]**2)

plt.figure(figsize=(12,4))

plt.plot(magnitude)

plt.title("Signal Magnitude")

plt.grid(True)

plt.show()
```
